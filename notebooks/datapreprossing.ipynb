```python
import os
import time

# drop any "%" characters that may be in the Technical competence % column
# (doing this before we coerce to numeric in the later cell)
df['Technical competence %'] = (
  df['Technical competence %']
    .astype(str)                            # make sure we have strings
    .str.replace('%', '', regex=False)      # drop any percent signs
    .str.replace(r'[^0-9.]', '', regex=True)  # strip out non‑numeric chars
    .str.strip()                            # trim surrounding whitespace
)
df['Technical competence %'].head(20)

# Save the changes to the CSV file
path = r'C:\Users\Ebrahim\Desktop\gurdproject\SmouhaMedicalCenter_cleaned.csv'

# Check if file exists and remove it first
if os.path.exists(path):
    try:
        os.remove(path)
        time.sleep(0.5)  # Wait a moment to ensure file is fully released
    except PermissionError:
        print(f"Warning: Could not remove {path}. File may be open in another application.")

df.to_csv(path, index=False)
print(f"File saved successfully to {path}")
```

```python
# Normalize 'Technical competence %' from 0-100 to 0-1
df['Technical competence %'] = df['Technical competence %'] / 100
```

```python
df['Technical competence %'].head(20)
```

```python
# if needed, ensure the columns are numeric (optional)
df['Technical competence %'] = pd.to_numeric(df['Technical competence %'], errors='coerce')
df['Functionality 100'] = pd.to_numeric(df['Functionality 100'], errors='coerce')

# fill the missing technical‑competence entries with the functionality score
df['Technical competence %'] = df['Technical competence %'].fillna(df['Functionality 100'])

# inspect a few rows to verify
df[['Technical competence %', 'Functionality 100']]
```
